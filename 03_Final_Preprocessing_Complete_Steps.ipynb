```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Load dataset
df = pd.read_csv("DATASET_PROJECT.csv")

# 1. Clean dataset: Remove duplicates and check nulls
df = df.drop_duplicates().reset_index(drop=True)
print(f"Total rows after removing duplicates: {len(df)}")
print(f"Missing values found: {df.isnull().sum().sum()}")

# 2. Separate Features (X) and Target (y)
X = df.drop(columns=['wine_type', 'quality'])
y = df['wine_type']

# 3. Stratified 80/20 Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# 4. Standard Scaling (Fit on Train, Transform on Test)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Convert back to DataFrame with feature names
X_train_processed = pd.DataFrame(X_train_scaled, columns=X.columns)
X_test_processed = pd.DataFrame(X_test_scaled, columns=X.columns)

# 5. Export processed splits to CSV
X_train_processed.to_csv("X_train_processed.csv", index=False)
X_test_processed.to_csv("X_test_processed.csv", index=False)
y_train.to_csv("y_train.csv", index=False)
y_test.to_csv("y_test.csv", index=False)

print("\nSuccessfully generated:")
print(" - X_train_processed.csv")
print(" - X_test_processed.csv")
print(" - y_train.csv")
print(" - y_test.csv")
```
