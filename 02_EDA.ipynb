```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load combined dataset
df = pd.read_csv("DATASET_PROJECT.csv")

# 1. Target Distribution Plot (Red vs. White)
plt.figure(figsize=(6, 4))
sns.countplot(x='wine_type', data=df, palette=['#F5A623', '#8B0000'])
plt.title("Distribution of Wine Types (0: White, 1: Red)")
plt.xlabel("Wine Type (0 = White, 1 = Red)")
plt.ylabel("Sample Count")
plt.show()

# 2. Correlation Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(df.corr(), annot=True, fmt=".2f", cmap="coolwarm", cbar=True)
plt.title("Physicochemical Features Correlation Heatmap")
plt.tight_layout()
plt.show()

# 3. Key Feature Comparisons by Wine Type
features = ['total sulfur dioxide', 'volatile acidity', 'chlorides', 'residual sugar']
plt.figure(figsize=(12, 8))
for i, col in enumerate(features, 1):
    plt.subplot(2, 2, i)
    sns.boxplot(x='wine_type', y=col, data=df, palette=['#F5A623', '#8B0000'])
    plt.title(f"{col} by Wine Type")
plt.tight_layout()
plt.show()
```
