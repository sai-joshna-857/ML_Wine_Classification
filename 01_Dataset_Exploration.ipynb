```python
import pandas as pd

# Load Red and White wine datasets directly from UCI
red_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv"
white_url = "https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-white.csv"

df_red = pd.read_csv(red_url, sep=';')
df_white = pd.read_csv(white_url, sep=';')

# Add binary target label: 1 = Red, 0 = White
df_red['wine_type'] = 1
df_white['wine_type'] = 0

# Merge into a single dataset
df = pd.concat([df_red, df_white], ignore_index=True)

# Check info & statistics
print("Dataset Shape:", df.shape)
print("\nClass Distribution:\n", df['wine_type'].value_counts())

# Save combined dataset to CSV
df.to_csv("DATASET_PROJECT.csv", index=False)
print("\nSaved to DATASET_PROJECT.csv successfully!")
```
